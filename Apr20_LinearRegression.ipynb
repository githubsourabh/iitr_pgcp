```python
#a = [5,2,2]
#b = [3, 2, 4]

def mean_squared_error(a, b):
    abs_error = 0
    for a1, b1 in zip(a, b):
        diff = a1 - b1
        abs_error += diff*diff
    
    return abs_error/len(a)
```

```python
def linear_reg_weights_bias(arr_of_x, arr_of_y):
    
    def find_starting_point(arr_of_x, arr_of_y):
        # Find the starting point for the weights and bias
        m = (arr_of_y[1]-arr_of_y[0])/(arr_of_x[1]-arr_of_x[0])
        c = arr_of_y[0] - m*arr_of_x[0]
        return m, c
    
    acceptable_error = 0.001
    max_iterations = 1000
    
    inferenced_y = []
    
    m, c = find_starting_point(arr_of_x, arr_of_y)

    inferenced_y.append(0)
    inferenced_y.append(0)
        
    for i in range (2,len(arr_of_x)):
        # inferencing the y value
        inferenced_y.append(m*arr_of_x[i] + c)

    # calculating the error
    error_prev = mean_squared_error(arr_of_y, inferenced_y)

    m_increment = 0.01
    
    iteration = 0
    
    while True:
        
        iteration += 1
        if iteration > max_iterations:
            break
        new_m = m + m_increment

        for i in range (len(arr_of_x)):
            # inferencing the y value
            inferenced_y[i] = new_m*arr_of_x[i] + c
        
        # calculating the error
        error_new = mean_squared_error(arr_of_y, inferenced_y)

        if (error_new < error_prev):
            m = new_m
            error_prev = error_new

        else:
            new_m = m - m_increment

            for i in range (len(arr_of_x)):
                # inferencing the y value
                inferenced_y[i] = new_m*arr_of_x[i] + c
            
            # calculating the error
            error_new = mean_squared_error(arr_of_y, inferenced_y)

            if (error_new < error_prev):
                m = new_m
                error_prev = error_new

            else:
                break

        if (abs(error_new-error_prev) < acceptable_error):
            break

    # calculating the bias
    for i in range (len(arr_of_x)):
        # inferencing the y value
        inferenced_y[i] = m*arr_of_x[i] + c

    error_prev = mean_squared_error(arr_of_y, inferenced_y)

    for i in range (len(arr_of_x)):
    
        bias_increment = 0.01
        iteration = 0


    while True:
        iteration += 1
        if iteration > max_iterations:
            break
        new_c = c + bias_increment

        for i in range (len(arr_of_x)):
            # inferencing the y value
            inferenced_y[i] = m*arr_of_x[i] + new_c
        
        # calculating the error
        error_new = mean_squared_error(arr_of_y, inferenced_y)

        if (error_new < error_prev):
            c = new_c
            error_prev = error_new

        else:
            new_c = c - bias_increment

            for i in range (len(arr_of_x)):
                # inferencing the y value
                inferenced_y[i] = m*arr_of_x[i] + new_c
            
            # calculating the error
            error_new = mean_squared_error(arr_of_y, inferenced_y)

            if (error_new < error_prev):
                c = new_c
                error_prev = error_new

            else:
                break

        if (abs(error_new-error_prev) < acceptable_error):
            break
    return m, c
```

```python
arr_of_x = [10,40,80]
arr_of_y = [8,12,18]
m, c = linear_reg_weights_bias(arr_of_x, arr_of_y)
print("m: ", m)
print("c: ", c)
inferenced_y = []
for i in range (len(arr_of_x)):
    # inferencing the y value
    inferenced_y.append(m*arr_of_x[i] + c)
print("inferenced_y: ", inferenced_y)
print("arr_of_y: ", arr_of_y)
print("Mean Squared Error: ", mean_squared_error(arr_of_y, inferenced_y))
```
