```python
#Return any base10 number in Base X (X is between 2 to 50)
SYMBOLS = "0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ"
def convertBase10toNewBase(base,num):
    arr=[]
    arr.clear()
    if base <2 or base >36:
        print ("base should be between 2 to 36")
    return convertBase10toNewBaseInternal(base,num,arr)

def convertBase10toNewBaseInternal(base,num,arr):
    new_num=num//base
    remainder = num%base
    if new_num !=0:
        convertBase10toNewBaseInternal(base,new_num,arr)
    arr.append(SYMBOLS[remainder])
    return arr
        
```

```python
#Convert any number of any base between 2 to 36, to Base 10
SYMBOLS = "0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ"
def convertNumberInGivenBasetoBase10(base,num):
    if base <2 or base >36:
        print ("base should be between 2 to 36")
    placeValue=0
    valueInDec=0
    while True: 
        remainder = num[-1]
        num=num[:-1]
        valueInDec+=((SYMBOLS.index(remainder))*(base**placeValue))
        placeValue+=1
        if num=='':
            break
    return valueInDec

```

```python
convertNumberInGivenBasetoBase10(20,"A12AAFEG")
```

```python
#Add two numbers in any base
def addTwoNumberInGivenBase(base,num1,num2):
    num1InBase10=convertNumberInGivenBasetoBase10(base,str(num1))
    num2InBase10=convertNumberInGivenBasetoBase10(base,str(num2))
    sumInBase10=num1InBase10+num2InBase10
    return convertBase10toNewBase(base,sumInBase10)
```

```python
addTwoNumberInGivenBase(2,101,110)
```

```python
#Substract two numbers in any base
def substractTwoNumberInGivenBase(base,num1,num2):
    num1InBase10=convertNumberInGivenBasetoBase10(base,str(num1))
    num2InBase10=convertNumberInGivenBasetoBase10(base,str(num2))
    numInBase10=num1InBase10-num2InBase10
    return convertBase10toNewBase(base,numInBase10)
```

```python
substractTwoNumberInGivenBase(2,11,1)
```
