```python
# Ques 1
def generate(n,m):
    arr = []
    for i in range (n):
        for j in range (m):
            arr.append((10**i)*j)
    return arr
```

```python
def predict (model, input):
    return (model[0]*input**2+(model[1]*input)+model[2])

#print (model)
predict (model, 3)
```
