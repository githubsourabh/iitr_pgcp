```python
# input is in the form of matrix
# for an equation ax + by =c  and a2x+b2y=c2 the matrix will be
# [a   b    c  ]
# [a2  b2   c2 ]
import copy
def solveLinearEquationRecursion(matrix):
    if len (matrix) == 1 :
        first_weight = []
        first_weight.append(matrix[0][1]/matrix[0][0])
        return (first_weight)
           
    #copy second last column of the matrix into a separate row. Second last column of the matrix is the last column before the "y" part which is the last column. All equations are in the form a(x1)+b(x2) + c(x3) =y
    second_last_column = [row[-2] for row in matrix]

    #print ("second last column is ", second_last_column)

    #print ("matrix before multiplication",matrix)

    for i in range ((len(matrix)-1)):
        for j in range (len (matrix[0])):
            matrix[i+1][j]=matrix[i+1][j]*second_last_column[0]
    #print (f"matrix after multiplying quoficient of second last element  of first row with all the rows ie ie {last_element_first_row} result {matrix}")
    
    
    for i in range ((len(matrix)-1)):
        for j in range (len (matrix[0])):
            matrix[i+1][j]=matrix[i+1][j]-matrix[0][j]*second_last_column[i+1]

    #print ("matrix after substracting first row multiplied by quoficieint of second last column of a particular row from that row ", matrix)

 
    # Remove the first row and second last column
    modified_matrix = [row[: -2] + row[-1:] for row in matrix[1:]]
    #print ("matrix after removing first row and second last column", modified_matrix)
    list_of_weights = solveLinearEquationRecursion (modified_matrix)
    #print (f"list_of_weights is of type {type (list_of_weights)} and value is {list_of_weights} and lenght is {len(list_of_weights)}")
    x =0
    for i in range (len(list_of_weights)):
    #    print (f"matrix[0] is {matrix[0]} and list_of_weights[i] is {list_of_weights[i]}")
        x += matrix[0][i]*list_of_weights[i]
    #    print (f"x is {x}")
    weight_found_this_iteration = (matrix[0][-1]-x)/matrix[0][-2]
    list_of_weights.append(weight_found_this_iteration)
    return list_of_weights


```
