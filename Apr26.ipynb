Q: Find the smallest value of a convex arra which first decreases and then increases:
arr = [10, 9, 8, 6.5, 4.1, 3.2, 2, 4, 4.5, 6]
Ans: 2

```python
def smallest_value_convex_arr(arr):
    left = 0
    right = len(arr) - 1
    mid = -1
    while left < right:
        mid = (left + right) // 2
        if arr[mid] < arr[mid -1] and arr[mid] < arr[mid + 1]:
            return arr[mid]
        elif arr[mid] > arr[mid - 1] and arr[mid] < arr[mid + 1]:
            right = mid 
        else:
            left = mid
    print (f"left: {left}, right: {right}, mid: {mid}, element: {arr[mid]}")

    if mid != -1:
        return arr[mid]
    else:
        return -1
    
```

```python
arr = [10, 9, 8, 6.5,6, 4.1, 4, 3.2, 2, 4, 4.5, 6,10,12]
smallest_value_convex_arr(arr)

```
