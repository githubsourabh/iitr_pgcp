```python
def sqrt(num):
    left = 0
    if num <1:
        right=1
    else:
        right = num
    while True:
        mid = (left+right)/2
        result = mid*mid
        diff = abs(result - num)
        if diff <= 0.0001:
            return (mid)
        else:
            if result > num:
                right = mid
            else:
                left = mid
                
    
```

```python
print (sqrt((0.4)))
```

```python
print (sqrt (5))
```

```python
def pythagoras (b,h):
    return sqrt((b*b)+(h*h))
```

```python
pythagoras (2,5)
```

```python
pythagoras (3,4)
```

```python
def distance_2d(x1, y1, x2, y2):
    return pythagoras (x1-x2,y1-y2)
```

```python
distance_2d(1,2,4,1)
```

```python
def distance_3d(x1, y1, z1, x2, y2, z2):
    dist_2d = distance_2d(x1,y1,x2,y2)
    return pythagoras (z1-z2,dist_2d)
```

```python
distance_3d (8,1,5,3,6,5)
```

```python
distance_3d (8,1,5,3,6,8)
```

```python
def differentialCalc(f,x):
    delta=0.000000000000001
    return (f(x+delta)-f(x))/delta
```
